```python
import polars as pl
```

```python
lf = pl.scan_csv('./dataset/train.csv', separator=';')
display(lf.head(10).collect())
```

```python
df = lf.filter(
    # pl.col('garage_number') == 31075,
    pl.col('ngpt_route') == '25 трамвай'
).with_columns(
    pl.col('tran_date_time').str.to_datetime(format="%Y-%m-%d %H:%M:%S")
).filter(
    pl.col('tran_date_time') >= pl.datetime(2025, 1, 1),
    pl.col('tran_date_time') < pl.datetime(2025, 1, 2)
).collect()

display(df)
```
